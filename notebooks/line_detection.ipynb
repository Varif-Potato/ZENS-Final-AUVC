```python
import os
os.environ["OPENCV_OPENCL_RUNTIME"] = ""
import cv2
import numpy as np 
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('lanes-15b8289a06a1b1e8a17347ad29bb1988.png', cv2.IMREAD_COLOR)
colorimg = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(colorimg)
```

```python
print(img.shape)
print(img is None)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) # convert to grayscale
plt.imshow(gray)
edges = cv2.Canny(gray, 10, 35, apertureSize=3) # detect edges
plt.imshow(edges)
```

```python
lines = cv2.HoughLinesP(
                edges,
                1,
                np.pi/180,
                threshold=55,
                minLineLength=250,
                maxLineGap=35,
        ) # detect lines

if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line
        cv2.line(img, (x1, y1), (x2, y2), (0, 255, 0), 5)
else:
    print("No lines detected")

plt.imshow(img)
```
